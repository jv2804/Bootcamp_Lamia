```python
import numpy as np
import pandas as pd
```

```python
data = {"Empresa": ["GOOG", "GOOG", "MSFT", "MSFT", "FB", "FB"],
        "Nome": ["Sam", "Charlie", "Amy", "Vanessa", "Carl", "Sarah"],
        "Venda": [200, 120, 340, 124, 245, 350]}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
#agrupa os valores não distintos em um só
group = df.groupby("Empresa")
group
```

```python
#soma todos os valores dentro do agrupamento
group.sum()
```

```python
#procura o valor minimo de cada grupo
group.min()
```

```python
#procura o valor maximo de cada grupo
group.max()
```

```python
#faz a media dos numeros(caso o numeric_only fosse False, daria erro, por haver divisao de str)
group.mean(numeric_only=True)
```

```python
#passa grande quantidade de informações importantes
group.describe()
```

```python
#mostra quantos dados existem dentro de cada grupo
group.count()
```

```python
#faz um agrupamento de valores sem similaridades, resultando em um df com o indice sendo o valor definido
group2 = df.groupby(by="Nome")
```
