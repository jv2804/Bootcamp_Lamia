```python
import numpy as np
import pandas as pd
```

```python
from numpy.random import randn
np.random.seed(101)
```

```python
df = pd.DataFrame(randn(5, 4), index="A B C D E".split(), columns= "W X Y Z".split())
```

```python
df
```

```python
#devolve booleanas para usar em analise de dados
df > 0
```

```python
bol = df > 0
```

```python
#ordena os valores por quais são verdadeiros e falsos
df[bol]
```

```python
#mesma coisa com o de cima, mas com outra condição
df[df["W"] > 0]
```

```python
#olha a coluna enquanto se aplica uma booleana para ordenacao
df[df["W"] > 0]["Y"]
```

```python
#mesmo processo do acima, mas com uma outra ordem
bol = df["W"] > 0
df2 = df[bol]
df2["Y"]
```

```python
#Nao funciona por causa do and
df[(df["W"] > 0) and (df["Y"] > 1)]
```

```python
#forma correta de se aplicar a comparação "e" entre duas booleanas
df[(df["W"] > 0) & (df["Y"] > 1)]
```

```python
#Nao funciona por causa do or
df[(df["W"] > 0) or (df["Y"] > 1)]
```

```python
#forma correta de se aplicar a comparação "or" entre duas booleanas
df[(df["W"] > 0) | (df["Y"] > 1)]
```

```python
#Caso o indice nao seja o padrão(numeros), ele transforma de volta para o padrão, transformando o indice antigo em uma coluna
df.reset_index(inplace=True)
```
