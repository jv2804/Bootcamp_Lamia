```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame({"col1": [1, 2, 3, 4],
                  "col2": [444, 555, 666, 444],
                  "col3": ["abc", "def", "ghi", "xyz"]})
df.head()
```

```python
#unique mostra os numeros distintos presentes na linha/coluna
df["col2"].unique()
```

```python
np.unique(df["col2"])
```

```python
#com isso podemos ver o numero de numeros distintos dentro da linha/coluna
len(df["col2"].unique())
```

```python
df["col2"].nunique()
```

```python
#com isso podemos ver ambos os acima, mostrando os numeros distintos e a quantidade de vezes que aparecem
df["col2"].value_counts()
```

```python
df[(df["col1"] > 2) & (df["col2"] == 444)]
```

```python
def vezes2(x):
    return x*2
```

```python
df["col1"].sum()
```

```python
#apply aplica funcoes em cada um dos elementos da Serie/DataFrame
df["col1"].apply(vezes2)
```

```python
#aplica as funcoes, mesmo as lambda criadas dentro do apply
df["col1"].apply(lambda x: x*x)
```

```python
#del deleta linhas/colunas
del(df["col2"])
```

```python
#ordena os valores da coluna/linha colocada no by, usa inplace como False
df.sort_values(by="col1")
```

```python
#verifica os valores nulos
df.isnull()
```

```python
df = pd.DataFrame({"col1": [1, 2, 3, np.nan],
                  "col2": [np.nan, 555, 666, 444],
                  "col3": ["abc", "def", "ghi", "xyz"]})
df.head()
```

```python
#inplace false por padrao
df["col1"].fillna(value=df["col1"].mean())
```

```python
data = {"A": ["foo", "foo", "foo", "bar", "bar", "bar"],
        "B": ["one", "one", "two", "two", "one", "one"],
        "C": ["x", "y", "x", "y", "x", "y"],
        "D": [1 , 3, 2, 5, 4, 1]}
df = pd.DataFrame(data)
```

```python
df
```

```python
#cria uma tabela dinâmica, transformando valores repetidos e não nulos em indices multi-niveis
df.pivot_table(values="D", index=["A", "B"], columns=["C"])
```
