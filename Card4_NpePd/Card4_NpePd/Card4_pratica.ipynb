```python
#crio dois arrays para criar unidimensionais para criar um DataFrame
arr1 = np.arange(0, 19789)
arr2 = np.arange(19789, 0, -1)
print(arr1)
print(arr2)
```

```python
ser1 = pd.Series(arr1)
ser2 = pd.Series(arr2)
ser1
```

```python
ser2
```

```python
#concateno as duas Series criadas em uma, usando o eixo x
ser3 = pd.concat([ser1, ser2], axis=1)
ser3
```

```python
#Crio um DataFrame apartir da Series criada anteriormente
df2 = pd.DataFrame(ser3)
df2
```

```python
#aplico uma função lambda em cada elemento dentro do DataFrame
df2.apply(lambda x: x*10)
```

```python
#operação para ver os elementos do DataFrame menores que 100
df2[df2 < 100]
```

```python
#faço a junção do df com o df2 com o método join
df.join(df2)
```

```python
#aqui tem bastante dados nulos
df["common_name"]
```

```python
#tratamento retirando colunas com dados nulos
df_op1 = df.dropna(axis=1)
df_op1
```

```python
#ao inves de remover, substitui o dado ausente com um valor padrão
df_op2 = df.fillna(value="Null")
df_op2["common_name"]
```

```python
#retiro uma das colunas criadas para a prática
df.drop("Coluna_Demo", axis=1)
```

```python
#coloquei o player_id como o indice
df.set_index(df["player_id"], inplace=True)
df
```

```python
#resetei o indice de volta ao padrão
df.reset_index(drop=True,inplace=True)
df
```

```python
#multiplos indices de maneira facilitada
df.set_index(["overall_rating", "player_id"])
```

```python
#agrupa os valores iguais de uma coluna
overall_df = df.groupby("overall_rating")
```

```python
#gera a maioria das estátisticas que possam ser interessantes
overall_df.describe()
```

```python
#operação booleana para ver bons jogadores atacantes
cond1 = df["shooting"] > 80
cond2 = df["dribbling"] > 85
df[cond1 & cond2]["overall_rating"]
```

```python
#verifica se o jogador eh bom em atirar ou em driblar
df[cond1 | cond2]["overall_rating"]
```
