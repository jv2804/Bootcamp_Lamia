```python
import numpy as np
import pandas as pd
```

```python
df1 = pd.DataFrame({"A": ["A0", "A1", "A2", "A3"], 
                   "B": ["B0", "B1", "B2", "B3"],
                   "C": ["C0", "C1", "C2", "C3"],
                   "D": ["D0", "D1", "D2", "D3"]
                   }, index=[0, 1, 2, 3])
df1
```

```python
df2 = pd.DataFrame({"A": ["A4", "A5", "A6", "A7"], 
                   "B": ["B4", "B5", "B6", "B7"],
                   "C": ["C4", "C5", "C6", "C7"],
                   "D": ["D4", "D5", "D6", "D7"]
                   }, index=[4, 5, 6, 7])
df2
```

```python
df3 = pd.DataFrame({"A": ["A8", "A9", "A10", "A11"], 
                   "B": ["B8", "B9", "B10", "B11"],
                   "C": ["C8", "C9", "C10", "C11"],
                   "D": ["D8", "D9", "D10", "D11"]
                   }, index=[8, 9, 10, 11])
df3
```

```python
#junta todos os DataFrames/Series em um unico DataFrame 
#agrupa os semelhantes em uma unica linha/coluna, os que nao se agrupam ficam em uma linha/coluna separada.
pd.concat([df1, df2, df3])
```

```python
cont = pd.concat([df1, df2, df3])
type(cont)
```

```python
#concatenacao entre linhas, como o axis eh por linhas, todos os DataFrames contatenados tem suas colunas separadas
pd.concat([df1, df2, df3], axis=1)
```

```python
esquerda = pd.DataFrame({ "key": ["K0", "K1", "K2", "K3"],
                          "A": ["A0", "A1", "A2", "A3"], 
                          "B": ["B0", "B1", "B2", "B3"]})
esquerda
```

```python
direita = pd.DataFrame({ "key": ["K0", "K1", "K2", "K3"],
                          "C": ["C0", "C1", "C2", "C3"],
                          "D": ["D0", "D1", "D2", "D3"]})
direita
```

```python
pd.concat([esquerda, direita])
```

```python
pd.concat([direita, esquerda])
```

```python
#mescla todos os DataFrames desde que contenham um elemento em comum(nesse exemplo, o "key")
#o atributo how define qual metodo sera usado para mesclagem, suas opcoes sao parecidas com SQL
pd.merge(esquerda, direita, how='inner' ,on='key')
```

```python
esquerda2 = pd.DataFrame({ "key": ["K0", "K0", "K1", "K2"],
                          "key2": ["K0", "K1", "K0", "K1"],
                          "A": ["A0", "A1", "A2", "A3"], 
                          "B": ["B0", "B1", "B2", "B3"]})

direita2 = pd.DataFrame({ "key": ["K0", "K0", "K1", "K2"],
                         "key2": ["K0", "K0", "K0", "K0"],
                          "C": ["C0", "C1", "C2", "C3"],
                          "D": ["D0", "D1", "D2", "D3"]})
```

```python
esquerda2
```

```python
direita2
```

```python
pd.merge(esquerda2, direita2,on=['key', 'key2'])
```

```python
#Usa a uniao de chaves de ambas os frames(esquerda2, direita2), seguindo o padrao de um dicionario(a-z,0-10)
pd.merge(esquerda2, direita2,how="outer",on=['key', 'key2'])
```

```python
#Usa a uniao de chaves do DataFrame a direita(nesse caso, direita2).
pd.merge(esquerda2, direita2,how="right",on=['key', 'key2'])
```

```python
#Usa a uniao de chaves do DataFrame a esquerda(nesse caso, esquerda2).
pd.merge(esquerda2, direita2,how="left",on=['key', 'key2'])
```

```python

esquerda3 = pd.DataFrame({ "A": ["A0", "A1", "A2"], 
                          "B": ["B0", "B1", "B2"]},
                        index=["K0", "K1", "K2"])

direita3 = pd.DataFrame({"C": ["C0", "C2", "C3"],
                        "D": ["D0", "D2", "D3"]},
                      index=["K0", "K2", "K3"])
```

```python
#mescla todos os DataFrames que podem nao conter um elemento em comum
#usa o objeto como referencia(esquerda3) para juntar o DataFrame inserido(direita3)
#coloca os valores do direita3 sempre que um dos indices dele corresponder com a referencia
esquerda3.join(direita3)
```

```python
direita3.join(esquerda3)
```

```python
direita3.join(esquerda3, how="outer")
```
