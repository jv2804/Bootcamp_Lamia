```python
#Enquanto o df segue sendo DataFrame
type(df)
```

```python
#Permite analisar um conjunto de colunas
df[["W", "Z"]]
```

```python
#não recomendado, pois pode ser confundido com uma função
df.W
```

```python
#tenta criar uma nova coluna sem dados, erro
df["NEW"]
```

```python
#cria uma nova coluna somando os dados de outras colunas
df["NEW"] = df["W"] + df["X"]
```

```python
df
```

```python
#remove a coluna criada apenas para exibição(implace tem que estar True para afetar a df diretamente)
df.drop("NEW", axis = 1)
```

```python
df
```

```python
#o que foi dito anteriormente eh aplicado aqui
df.drop("NEW", axis = 1, inplace=True)
```
