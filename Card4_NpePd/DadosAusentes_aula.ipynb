```python
import numpy as np
import pandas as pd
```

```python
d = {"A": [1, 2, np.nan], "B": [5, np.nan, np.nan], "C": [1, 2, 3]}
```

```python
df = pd.DataFrame(d)
```

```python
df
```

```python
#retira as linhas/colunas com valores nulos
df.dropna()
```

```python
#remove apenas linhas/colunas com mais de 2 nulos
df.dropna(thresh = 2)
```

```python
#preenche os dados nulos com o valor definido
df.fillna(value="fill na")
```

```python
#preenche os dados nulos na coluna A com a media dos valores antecedores 
df["A"].fillna(value=df["A"].mean())
```

```python
#No video, ffill era um das opcoes de um atributo do fillna chamado "method", agora ffill eh um metodo separado
#preenche os dados nulos com valores que o antecedem bom pra ser usado com analise em tempos reais
df.ffill()
```

```python
#aplica o dropna no df
df.dropna(inplace = True)
```
