```python
import numpy as np
import pandas as pd
```

```python
#Criacao das labels
indice = ["Fruta", "Preco/KG", "Estoque", "Fornecedor", "Validade"]
indice
```

```python
#criacao dos dados
dados = ["Abacate", 4, 20, "Coppercitrus", 2], ["Banana", 6.99, 10, "Bananasportao", 3], ["Laranja", 1.78, 40.8, "Citropack", 5]
```

```python
#Labels colocadas em uma Serie
ser0 = pd.Series(indice)
ser0
```

```python
#DataFrame criado atraves dos dados junto das labels
df = pd.DataFrame(dados,columns=ser0)
#coloca Fruta como indice
df.set_index("Fruta", inplace=True)
df
```

```python
#cria uma nova coluna que é produto do preço com a quantidade no estoque
df["PrecoTotal"] = df["Preco/KG"]*df["Estoque"]
df
```

```python
#localiza uma unica linha e passa por ela como uma Serie
df.loc["Banana"]
```

```python
#criacao de um segundo DataFrame para concatenar
df2 = pd.DataFrame({"Fruta": ["Amora"],
                    "Preco/KG": [np.nan],
                   "Estoque": [np.nan],
                   "Fornecedor": [np.nan],
                   "Validade": [np.nan],
                   "PrecoTotal": [np.nan]})
df2.set_index("Fruta", inplace=True)
df2
```

```python
#cria sem querer duas instancias da df2 dentro da df
df = pd.concat([df,df2])
df = pd.concat([df,df2])
df
```

```python
#deleta todas as instancias da df2
df.drop(labels="Amora", inplace=True)
df
```

```python
#corrige o erro, fazendo apenas uma unica instancia
df = pd.concat([df,df2])
df
```

```python
#enche a parte null da Amora, Preco/KG com um valor especifico
df.fillna({"Preco/KG": 30}, inplace=True)
```

```python
#modifica o nome da coluna Estoque para fazer mais sentido
df.rename(columns={"Estoque": "Estoque(KG)"}, inplace=True)
```

```python
#enche as partes nulas da Amora com dados da Laranja
df.ffill(inplace = True)
```
