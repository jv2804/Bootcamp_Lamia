# Exercise: Mean & Median Customer Spend

Here's some code that will generate some random e-commerce data; just an array of total amount spent per transaction. Select the code block, and hit "play" to execute it:

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

incomes = np.random.normal(100.0, 20.0, 10000)

plt.hist(incomes, 50)
plt.show()
```

Now, find the mean and median of this data. In the code block below, write your code, and see if your result makes sense:

```python
#exercicio pede para achar a media e mediana da data acima, além de verificar se bate com o resultado acima.
np.mean(incomes)
```

```python
np.median(incomes)
```

This is pretty much the world's easiest assignment, but we're just trying to get your hands on iPython and writing code with numpy to get you comfortable with it.

Try playing with the code above to generate different distributions of data, or add outliers to it to see their effect.

```python
#aqui ele pede para alterar os dados um pouco, vou adicionar 2 outliers, um com valor muito menor do q a média e um muito maior
incomes = np.append(incomes, [-100, 10000000000])
```

```python
#como podemos ver, a média se alterou muito
np.mean(incomes)
```

```python
#enquanto a mediana se manteve igual(se deve ao fato de ter sido adicionado 2 valores, 
#caso apenas um valor tivesse sido introduzido, a mediana seria alterada por pouco)
np.median(incomes)
```
