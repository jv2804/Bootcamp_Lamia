# Percentiles

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

vals = np.random.normal(0, 0.5, 10000)

plt.hist(vals, 50)
plt.show()
```

```python
np.percentile(vals, 50)
```

```python
np.percentile(vals, 90)
```

```python
np.percentile(vals, 20)
```

## Activity

Experiment with different parameters when creating the test data. What effect does it have on the percentiles?

```python
#alterei o desvio padrão dos dados para 0.99
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

vals2 = np.random.normal(0, 0.99, 10000)

plt.hist(vals2, 50)
plt.show()
```

```python
#std mudado para 0.99
#o anterior estava com 0.5
vals2.std()
```

```python
vals.std()
```

```python
#como o ponto central continua o mesmo, 50 percentil permanece próximo do anterior
np.percentile(vals2, 50)
```

```python
#graças ao std ter quase dobrado, 90 percentil de agora é quase o dobro do 90 percentil do anterior
np.percentile(vals2, 90)
```

```python
#mesma coisa que ocorreu com o 90 percentil aconteceu com o 20 percentil
np.percentile(vals2, 20)
```
