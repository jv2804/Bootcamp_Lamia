# Standard Deviation and Variance

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

incomes = np.random.normal(100.0, 50.0, 10000)

plt.hist(incomes, 50)
plt.show()
```

```python
incomes.std()
```

```python
incomes.var()
```

Experiment with different parameters on the normal function, and see what effect it has on the shape of the distribution. How does that new shape relate to the standard deviation and variance?

```python
#é pedido para experimentar com diferentes parametros da função inicial, e ver o efeito na variância e no desvio padrão
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
#o primeiro valor influencia aonde será o ponto de maior qntd, o segundo valor afeta a variação e o desvio padrão
incomes2 = np.random.normal(314, 67, 100000)

plt.hist(incomes2, 50)
plt.show()
```
