# Moments: Mean, Variance, Skew, Kurtosis

Create a roughly normal-distributed random set of data:

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

vals = np.random.normal(0, 0.5, 10000)

plt.hist(vals, 50)
plt.show()
```

The first moment is the mean; this data should average out to about 0:

```python
np.mean(vals)
```

The second moment is the variance:

```python
np.var(vals)
```

The third moment is skew - since our data is nicely centered around 0, it should be almost 0:

```python
import scipy.stats as sp
sp.skew(vals)
```

The fourth moment is "kurtosis", which describes the shape of the tail. For a normal distribution, this is 0:

```python
sp.kurtosis(vals)
```

## Activity

Understanding skew: change the normal distribution to be centered around 10 instead of 0, and see what effect that has on the moments.

The skew is still near zero; skew is associated with the shape of the distribution, not its actual offset in X.

```python
#aqui ele pede para alterar a distribuição normal para ficar centralizado no 10 ao invés do 0, e ver qual efeito isso proporciona

%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

vals2 = np.random.normal(10, 0.5, 10000)

plt.hist(vals2, 50)
plt.show()
```

```python
#houve uma alteração pois a distribuição agora está centrada no 10
np.mean(vals2)
```

```python
#aqui podemos ver que a diferença da variancia foi mínima, isso ocorre pois os elementos estão igualmentes distribuidos
np.var(vals2)
```

```python
#como pode ser visto, a assimétria(skewness) é baixa, isso ocorre pelo mesmo motivo que a variancia
sp.skew(vals2)
```

```python
#mesma história que a variancia e a assimétria(skewness)
sp.kurtosis(vals2)
```
